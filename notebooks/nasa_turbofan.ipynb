# NASA Turbofan Engine Degradation Simulation (C-MAPSS)

Utvärdering av datasetet inför gruppens val. Ingen analys är gjord ännu.

## Datasetet
- Simulerad data från NASA:s motorsimulator C-MAPSS, där flygmotorer körs från start till haveri.
- Varje motor följs cykel för cykel med ett antal sensorvärden och driftinställningar.
- Den klassiska versionen består av fyra delmängder (FD001–FD004) med olika driftförhållanden och felorsaker, samt testdata med facit för återstående livslängd.
- **Kaggle:** https://www.kaggle.com/datasets/bishals098/nasa-turbofan-engine-degradation-simulation
- **Ursprunglig källa:** NASA Prognostics Center of Excellence. Beskrivs i Saxena m.fl. (2008), *Damage Propagation Modeling for Aircraft Engine Run-to-Failure Simulation*.

## Möjligt problem och användare
- **Problem:** Förutsäga återstående livslängd (Remaining Useful Life, RUL) för en motor utifrån sensordata.
- **Användare:** Underhållsplanerare som vill schemalägga service innan haveri, i stället för enligt fasta intervall eller efter att fel uppstått.
- Till skillnad från AI4I är detta prediktivt underhåll i verklig mening: en prognos framåt i tiden.

## Möjlig baseline och modeller
- **Baseline:** Medelvärdet av livslängden, eller en enkel linjär regression på antal körda cykler.
- **Modeller:** Random Forest, gradient boosting med variabler baserade på glidande medelvärden och trender, eventuellt en enkel LSTM som fördjupning.
- **Mått:** RMSE på testdatans facit, som är standard för datasetet och gör det lätt att jämföra med publicerade resultat.

## Styrkor
- Riktig tidsserie med tydlig nedbrytning mot haveri.
- Etablerat benchmark med färdig testdata och facit.
- Bra lärdom om dataläckage: uppdelningen i tränings- och valideringsdata måste göras **per motor**, inte per rad.

## Svagheter och risker
- Simulerad, inte verklig, data.
- Mycket använt dataset, så det finns många färdiga lösningar och det är svårare att tillföra något nytt.
- Sensorerna saknar beskrivande namn, vilket begränsar domänförståelsen.
- **Licens och tillgänglighet är oklar:**
  - Kaggle-versionen anges ha licensen *US Government Works* (allmän egendom i USA).
  - På data.gov anges att NASA för närvarande ser över tillgängligheten för C-MAPSS-mjukvaran, och en närliggande katalogpost är markerad som *non-public*.
  - Jag har inte kunnat bekräfta en officiell nedladdningssida med tydlig licens för den klassiska datamängden.
  - Kursen kräver att vi kan visa att datan får användas och att projektet inte beror på en osäker datakälla.
